```julia
# 1. Строка → целое
parse(Int, "123")               # → 123
parse(Int64, "42")              # → 42

# 2. Строка → вещественное
parse(Float64, "3.14")          # → 3.14
parse(Float32, "2.71")          # → 2.71f0

# 3. Строка → Bool
parse(Bool, "true")             # → true
parse(Bool, "false")            # → false

# 4. Система счисления
parse(Int, "ff", base=16)       # → 255
parse(Int, "1010", base=2)      # → 10

# 5. Комплексные числа
parse(Complex{Float64}, "1+2im")  # → 1.0 + 2.0im

# 6. Ошибка при неверном вводе
try
    parse(Int, "abc")
catch e
    println("Ошибка: ", e)
end

# 7. tryparse — безопасный вариант
tryparse(Int, "abc")            # → nothing
tryparse(Int, "42")             # → 42
```

```julia
println("Сложение:       2 + 3      = ", 2 + 3)
println("Вычитание:      5 - 2      = ", 5 - 2)
println("Умножение:      4 * 3      = ", 4 * 3)
println("Деление:        10 / 3     = ", 10 / 3)
println("Целочисл. дел.: 10 ÷ 3     = ", 10 ÷ 3)
println("Остаток:        10 % 3     = ", 10 % 3)
println("Степень:        2 ^ 10     = ", 2 ^ 10)
println("Квадр. корень:  sqrt(16)   = ", sqrt(16))
println("Куб. корень:    cbrt(27)   = ", cbrt(27))
println("Модуль:         abs(-7)    = ", abs(-7))
```

```julia
a = 5           # Int64
b = 2.5         # Float64
c = 3 + 4im     # Complex{Int64}

println("a + b = ", a + b, "  тип: ", typeof(a + b))
println("a * c = ", a * c, "  тип: ", typeof(a * c))
println("b / c = ", b / c, "  тип: ", typeof(b / c))
println("2^0.5 = ", 2^0.5, "  тип: ", typeof(2^0.5))
```

```julia
println("1 == 1.0   → ", 1 == 1.0)     # true (значения равны)
println("1 === 1.0  → ", 1 === 1.0)    # false (разные типы)
println("1 != 2     → ", 1 != 2)       # true
println("3 > 2      → ", 3 > 2)        # true
println("3 < 2      → ", 3 < 2)        # false
println("3 >= 3     → ", 3 >= 3)       # true
println("3 <= 2     → ", 3 <= 2)       # false
println("NaN == NaN → ", NaN == NaN)   # false
println("NaN != NaN → ", NaN != NaN)   # true
```

```julia
println("true && false → ", true && false)   # false
println("true || false → ", true || false)   # true
println("!true         → ", !true)           # false
println("xor(true,false) → ", xor(true, false))  # true

# Короткое замыкание
x = 5
println("x > 0 && x < 10 → ", x > 0 && x < 10)  # true
println("x < 0 || x > 3  → ", x < 0 || x > 3)   # true
```

```julia
A = [1 2; 3 4]
B = [5 6; 7 8]

println("A = ", A)
println("B = ", B)
println("A + B = ", A + B)
println("A - B = ", A - B)
```

```julia
println("2 * A = ", 2 * A)
println("A * 3 = ", A * 3)
println("A / 2 = ", A / 2)
```

```julia
println("A * B = ", A * B)
v = [1, 2]
println("A * v = ", A * v)
```

```julia
println("A' = ", A')          # транспонирование
println("v' = ", v')          # вектор-столбец → вектор-строка
```

```julia
v = [1, 2, 3]
w = [4, 5, 6]

# Способ 1: через транспонирование
println("v' * w = ", v' * w)

# Способ 2: функция dot
using LinearAlgebra
println("dot(v, w) = ", dot(v, w))
```
